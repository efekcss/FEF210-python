## Demet(Tuple)
Bir tuple sabit uzunlukta(fixed-length),değiştirilemez(immutable) bir python nesneleri sekansıdır. Bir demet oluşturmanın en kolay yolu, virgülle ayrılmış bir değerler sekansı girmektir. Bir dizi virgülle ayrılımş ifade, parantez verilmese bile bunlar tek bir demet olarak ele alınır.

```python
tup = 7, 8, 9

type(tup)
```

Ancak kimi zaman sekansın başına ve sonuna parantez konabilir:

```python
tup2 = (1,2,3,4)
print(tup2)
type(tup2)
```

```python
#boş bir demet oluşturma
s = ()
s = tuple() # tuple() fonksiyonu yerleşik fonksiyondur
```

Bir demetin elemanlarına, diğer birçok sekans türünde olduğu gibi köşeli parantez `[]` ile indeksi ile erişilebilir.

```python
tup3 = tuple("deneme")
tup3
```

```python
tup3[3]
```

**NOT**: C, C++, Java ve diğer birçok dilde olduğu gibi, Python'da indeksleme 0 ile başlar. Yani 0'ıncı indeksteki eleman, birinci öğedir:

```python
tup3[0]
```

Bir demet değiştirilemez (immutable) bir nesnedir. Bu nedenle, bir demet oluşturulduktan sonra, o demetteki nesneleri değiştiremezsiniz.

```python
tup4 = ('text', [1,3], True, None, 6)
```

```python
tup4[3] = 5
```

Bir demet içerisindeki bir elemanda, değiştirilebilir bir nesne bulunuyorsa (örneğin bir ;liste) o elemandaki nesneyi değiştirebilirsiniz:

```python
tup4[1].append(2)
```

```python
tup4
```

Daha uzun demetler oluşturmak için `+` operatörünü kullanarak demetleri birleştirebiliriz.


```python
print(tup)
print(tup2)
print(tup + tup2)
```

Lİstelerde olduğu gibi bir demeti bir tamsayı ile çarpmak , demetin o kadar çok kopyasını bir araya getirme etkisine sahiptir.

```python
tup8 = tup*4
print(tup8)
```

### Nesneleri Açma (unpack)

Bir demet-benzeri bir değişken ifadesi atamaya çalışırsanız, Python eşittir işaretinin sağ tarafındaki değeri açmaya çalışır:

```python
tup = (4.5e-16, 6, 7)

x, y, z = tup
```

```python
print(x)
print(y)
print(z)
```

İç içe demetlere (nested tuples) sahip sekanslar bile açılabilir:

```python
tup = (1,2,3,(4,5),6)
```

```python
a,b,c,(d,e),f = tup
```

```python
print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

Değişken açmanın yaygın bir kullanımı da demetlerin veya listelerin bir sekansı üzerinde yineleme (iteration) yapmaktır:

```python
# List of tuples (demetlerden oluşan bir liste)
sequence = [(1,2,3),(4,5,6),(7,8,9)]
```

```python
for i in sequence:
    print(i)
```

### Demet Metotları

Bir demetin boyutu ve içerği değiştirilemediğinden, bu yerleşik veri yapısının çok fazla metodu yoktur. Özellikle yararlı olan (listelerde de bulunur) bir değerin oluşum sayısını sayan `count` işlevidir:

```python
a = (1,2,2,2,3,4,5,2)
```

```python
a.count(2)
```

```python
a.index(5)
```

## Liste (List)

Demet'lerin aksine, listeler değişken uzunluktadır(variable-length) ve içerikleri yerinde(in-place) değiştirilebilir (yani bu veri yapısı mutable -değiştirilebilir veri yapısıdır-). Bir listeyi, köşeli parantez `[]` kullanarak veya `list` fonksiyonu kullanarak tanımlayabilirsiniz:

```python
a_list = [2, 4, 3, None]
```

ya da bir demeti bir listeye çevirebilirsiniz:

```python
tup = ('foo', 'dbar', 'baz')
```

```python
tup
```

```python
b_list = list(tup)
b_list
```

```python
b_list[1] = 'peekaboo'
```

```python
b_list
```

Listeler ve demetler anlamsal olara benzerdir ancak, demetler değiştirilemez ve birçok fonksiyonda birbirinin yerine kullanılabilir.
